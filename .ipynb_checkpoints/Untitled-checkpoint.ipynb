```python
import tkinter as tk
from tkinter import messagebox, simpledialog

DEFAULT_ROWS = 12
DEFAULT_COLS = 12
CELL_SIZE    = 38
PADDING      = 2

COLORS = {
    "empty":    "#1e1e2e",
    "start":    "#fab387",
    "end":      "#f38ba8",
    "obstacle": "#45475a",
    "grid_bg":  "#11111b",
    "text":     "#cdd6f4",
    "panel_bg": "#181825",
    "btn_bg":   "#313244",
    "btn_act":  "#89b4fa",
    "btn_text": "#cdd6f4",
    "border":   "#6c7086",
    "queue":    "#f9e2af",
    "visited":  "#89b4fa",
    "path":     "#a6e3a1",
    "agent":    "#cba6f7",
    "dynamic_obstacle": "#f38ba8",
}

MODE_INFO = {
    "start":    ("🟢 Start",    COLORS["start"],    "Place the START point"),
    "end":      ("🔴 End",      COLORS["end"],      "Place the END point"),
    "obstacle": ("⬛ Obstacle", COLORS["obstacle"], "Draw OBSTACLE walls"),
    "clear":    ("✖ Erase",    COLORS["empty"],    "Erase cells"),
}

class GridApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Informed Search Visualizer")
        self.configure(bg=COLORS["grid_bg"])
        self.resizable(False, False)

        self.rows    = DEFAULT_ROWS
        self.cols    = DEFAULT_COLS
        self.mode    = tk.StringVar(value="start")
        self.is_drag = False
        self.cells   = {}
        self.rects   = {}
        self.start   = None
        self.end     = None
        
        self._sim_running  = False
        self._sim_after_id = None
        self._overlays     = set()
        self._speed_var    = tk.IntVar(value=50)
        
        self._current_path = []
        self._current_pos_index = 0
        self._replan_count = 0

        self._build_ui()
        self._init_grid()

    def _build_ui(self):
        body = tk.Frame(self, bg=COLORS["grid_bg"])
        body.pack(padx=14, pady=10)

        self._build_left_panel(body)
        self._build_canvas(body)
        self._build_right_panel(body)

    def _build_left_panel(self, parent):
        panel = tk.Frame(parent, bg=COLORS["panel_bg"], padx=12, pady=12,
                         bd=1, relief="flat")
        panel.pack(side="left", fill="y", padx=(0, 12))

        self.mode_buttons = {}
        for key, (label, color, tip) in MODE_INFO.items():
            self._make_mode_btn(panel, key, label, color, tip)

        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=14)

        self._heuristic_var = tk.StringVar(value="manhattan")
        
        heur_frame1 = tk.Frame(panel, bg=COLORS["panel_bg"])
        heur_frame1.pack(fill="x", pady=3)
        tk.Radiobutton(
            heur_frame1, text="Manhattan Distance",
            variable=self._heuristic_var, value="manhattan",
            font=("Segoe UI", 9),
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            selectcolor=COLORS["btn_bg"],
            activebackground=COLORS["panel_bg"],
            activeforeground=COLORS["btn_act"],
            cursor="hand2"
        ).pack(anchor="w")
        
        heur_frame2 = tk.Frame(panel, bg=COLORS["panel_bg"])
        heur_frame2.pack(fill="x", pady=3)
        tk.Radiobutton(
            heur_frame2, text="Euclidean Distance",
            variable=self._heuristic_var, value="euclidean",
            font=("Segoe UI", 9),
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            selectcolor=COLORS["btn_bg"],
            activebackground=COLORS["panel_bg"],
            activeforeground=COLORS["btn_act"],
            cursor="hand2"
        ).pack(anchor="w")

        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=14)

        self._algo_var = tk.StringVar(value="best_first")
        
        algo_frame1 = tk.Frame(panel, bg=COLORS["panel_bg"])
        algo_frame1.pack(fill="x", pady=3)
        tk.Radiobutton(
            algo_frame1, text="Best-First Search",
            variable=self._algo_var, value="best_first",
            font=("Segoe UI", 9),
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            selectcolor=COLORS["btn_bg"],
            activebackground=COLORS["panel_bg"],
            activeforeground=COLORS["btn_act"],
            cursor="hand2"
        ).pack(anchor="w")
        
        algo_frame2 = tk.Frame(panel, bg=COLORS["panel_bg"])
        algo_frame2.pack(fill="x", pady=3)
        tk.Radiobutton(
            algo_frame2, text="A* Search",
            variable=self._algo_var, value="astar",
            font=("Segoe UI", 9),
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            selectcolor=COLORS["btn_bg"],
            activebackground=COLORS["panel_bg"],
            activeforeground=COLORS["btn_act"],
            cursor="hand2"
        ).pack(anchor="w")

        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=14)

        self._search_btn = tk.Button(
            panel, text="▶  Run Search",
            font=("Segoe UI", 10, "bold"),
            bg="#a6e3a1", fg="#11111b",
            activebackground="#94e2d5", activeforeground="#11111b",
            relief="flat", bd=0, padx=10, pady=7, anchor="w",
            cursor="hand2", command=self._run_search
        )
        self._search_btn.pack(fill="x", pady=3)

        self._stop_btn = tk.Button(
            panel, text="⏹  Stop",
            font=("Segoe UI", 10),
            bg=COLORS["btn_bg"], fg=COLORS["btn_text"],
            activebackground="#f38ba8", activeforeground="#11111b",
            relief="flat", bd=0, padx=10, pady=6, anchor="w",
            cursor="hand2", command=self._stop_sim, state="disabled"
        )
        self._stop_btn.pack(fill="x", pady=3)

        self._clear_path_btn = tk.Button(
            panel, text="🧹  Clear Path",
            font=("Segoe UI", 10),
            bg=COLORS["btn_bg"], fg=COLORS["btn_text"],
            activebackground="#cba6f7", activeforeground="#11111b",
            relief="flat", bd=0, padx=10, pady=6, anchor="w",
            cursor="hand2", command=self._clear_overlay
        )
        self._clear_path_btn.pack(fill="x", pady=3)

        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=10)

        speed_row = tk.Frame(panel, bg=COLORS["panel_bg"])
        speed_row.pack(fill="x")
        tk.Label(speed_row, text="Fast", font=("Segoe UI", 8),
                 bg=COLORS["panel_bg"], fg=COLORS["border"]).pack(side="left")
        tk.Scale(
            speed_row, from_=1, to=300,
            orient="horizontal", variable=self._speed_var,
            bg=COLORS["panel_bg"], fg=COLORS["btn_text"],
            troughcolor=COLORS["btn_bg"], highlightthickness=0,
            showvalue=False, sliderlength=14
        ).pack(side="left", fill="x", expand=True, padx=4)
        tk.Label(speed_row, text="Slow", font=("Segoe UI", 8),
                 bg=COLORS["panel_bg"], fg=COLORS["border"]).pack(side="left")

    def _make_mode_btn(self, parent, key, label, color, tip):
        frame = tk.Frame(parent, bg=COLORS["panel_bg"])
        frame.pack(fill="x", pady=3)

        dot = tk.Label(frame, text="●", font=("Segoe UI", 12),
                       bg=COLORS["panel_bg"], fg=color, width=2)
        dot.pack(side="left")

        btn = tk.Button(
            frame, text=label,
            font=("Segoe UI", 10),
            bg=COLORS["btn_bg"], fg=COLORS["btn_text"],
            activebackground=COLORS["btn_act"], activeforeground="#11111b",
            relief="flat", bd=0, padx=10, pady=6, anchor="w",
            cursor="hand2",
            command=lambda k=key: self._set_mode(k)
        )
        btn.pack(side="left", fill="x", expand=True)

        tk.Label(frame, text=tip, font=("Segoe UI", 7),
                 bg=COLORS["panel_bg"], fg=COLORS["border"]).pack(
                     side="left", padx=(4, 0))

        self.mode_buttons[key] = btn
        self._update_btn_style(key)

    def _make_action_btn(self, parent, text, cmd):
        btn = tk.Button(
            parent, text=text,
            font=("Segoe UI", 10),
            bg=COLORS["btn_bg"], fg=COLORS["btn_text"],
            activebackground="#f5c2e7", activeforeground="#11111b",
            relief="flat", bd=0, padx=10, pady=6, anchor="w",
            cursor="hand2", command=cmd
        )
        btn.pack(fill="x", pady=3)

    def _build_canvas(self, parent):
        canvas_w = self.cols * (CELL_SIZE + PADDING) + PADDING
        canvas_h = self.rows * (CELL_SIZE + PADDING) + PADDING

        frame = tk.Frame(parent, bg=COLORS["grid_bg"],
                         bd=2, relief="flat",
                         highlightbackground=COLORS["border"],
                         highlightthickness=1)
        frame.pack(side="left")

        self.canvas = tk.Canvas(
            frame, width=canvas_w, height=canvas_h,
            bg=COLORS["grid_bg"], highlightthickness=0, cursor="crosshair"
        )
        self.canvas.pack()

        self.canvas.bind("<Button-1>",        self._on_click)
        self.canvas.bind("<B1-Motion>",       self._on_drag)
        self.canvas.bind("<ButtonRelease-1>", lambda e: setattr(self, "is_drag", False))
        self.canvas.bind("<Button-3>",        self._on_right_click)
        self.canvas.bind("<B3-Motion>",       self._on_right_drag)

    def _build_right_panel(self, parent):
        panel = tk.Frame(parent, bg=COLORS["panel_bg"], padx=12, pady=12,
                         bd=1, relief="flat")
        panel.pack(side="left", fill="y", padx=(12, 0))

        self._make_action_btn(panel, "🔄  Clear Grid",    self._clear_grid)
        self._make_action_btn(panel, "↩  Clear Obstacles", self._clear_obstacles)
        
        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=10)
        
        size_frame = tk.Frame(panel, bg=COLORS["panel_bg"])
        size_frame.pack(fill="x", pady=(0, 8))
        
        tk.Label(size_frame, text="Rows:", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")
        self._rows_var = tk.StringVar(value="12")
        tk.Entry(size_frame, textvariable=self._rows_var, font=("Segoe UI", 9),
                 bg=COLORS["btn_bg"], fg=COLORS["text"], width=4,
                 insertbackground=COLORS["text"], relief="flat").pack(side="left", padx=(5, 10))
        
        tk.Label(size_frame, text="Cols:", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")
        self._cols_var = tk.StringVar(value="12")
        tk.Entry(size_frame, textvariable=self._cols_var, font=("Segoe UI", 9),
                 bg=COLORS["btn_bg"], fg=COLORS["text"], width=4,
                 insertbackground=COLORS["text"], relief="flat").pack(side="left", padx=5)
        
        self._make_action_btn(panel, "⚙  Apply Size", self._resize_grid)
        
        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=10)
        
        density_frame = tk.Frame(panel, bg=COLORS["panel_bg"])
        density_frame.pack(fill="x", pady=(0, 8))
        tk.Label(density_frame, text="Density:", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")
        self._density_var = tk.IntVar(value=30)
        tk.Label(density_frame, textvariable=self._density_var, font=("Segoe UI", 9, "bold"),
                 bg=COLORS["panel_bg"], fg=COLORS["btn_act"], width=3).pack(side="left", padx=5)
        tk.Label(density_frame, text="%", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")
        
        tk.Scale(
            panel, from_=0, to=80,
            orient="horizontal", variable=self._density_var,
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            troughcolor=COLORS["btn_bg"], highlightthickness=0,
            showvalue=False, sliderlength=14
        ).pack(fill="x", pady=(0, 8))
        
        self._make_action_btn(panel, "🎲  Generate Maze", self._generate_maze)

        tk.Frame(panel, bg=COLORS["border"], height=1).pack(fill="x", pady=10)

        self._dynamic_var = tk.BooleanVar(value=False)
        tk.Checkbutton(
            panel, text="🔄 Dynamic Mode",
            variable=self._dynamic_var,
            font=("Segoe UI", 10, "bold"),
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            selectcolor=COLORS["btn_bg"],
            activebackground=COLORS["panel_bg"],
            activeforeground=COLORS["btn_act"],
            cursor="hand2"
        ).pack(anchor="w", pady=(0, 8))

        spawn_frame = tk.Frame(panel, bg=COLORS["panel_bg"])
        spawn_frame.pack(fill="x", pady=(0, 8))
        tk.Label(spawn_frame, text="Spawn:", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")
        self._spawn_prob_var = tk.IntVar(value=5)
        tk.Label(spawn_frame, textvariable=self._spawn_prob_var, font=("Segoe UI", 9, "bold"),
                 bg=COLORS["panel_bg"], fg=COLORS["btn_act"], width=3).pack(side="left", padx=5)
        tk.Label(spawn_frame, text="%", font=("Segoe UI", 9),
                 bg=COLORS["panel_bg"], fg=COLORS["text"]).pack(side="left")

        tk.Scale(
            panel, from_=1, to=20,
            orient="horizontal", variable=self._spawn_prob_var,
            bg=COLORS["panel_bg"], fg=COLORS["text"],
            troughcolor=COLORS["btn_bg"], highlightthickness=0,
            showvalue=False, sliderlength=14
        ).pack(fill="x")


        self.bind("<KeyPress-s>", lambda e: self._set_mode("start"))
        self.bind("<KeyPress-S>", lambda e: self._set_mode("start"))
        self.bind("<KeyPress-e>", lambda e: self._set_mode("end"))
        self.bind("<KeyPress-E>", lambda e: self._set_mode("end"))
        self.bind("<KeyPress-o>", lambda e: self._set_mode("obstacle"))
        self.bind("<KeyPress-O>", lambda e: self._set_mode("obstacle"))
        self.bind("<KeyPress-x>", lambda e: self._set_mode("clear"))
        self.bind("<KeyPress-X>", lambda e: self._set_mode("clear"))
        self.bind("<Delete>",     lambda e: self._clear_grid())
        self.bind("<BackSpace>",  lambda e: self._clear_grid())

    def _init_grid(self):
        self._stop_sim()
        self.canvas.delete("all")
        self.cells  = {}
        self.rects  = {}
        self.start  = None
        self.end    = None
        self._overlays = set()

        for r in range(self.rows):
            for c in range(self.cols):
                x1 = c * (CELL_SIZE + PADDING) + PADDING
                y1 = r * (CELL_SIZE + PADDING) + PADDING
                x2 = x1 + CELL_SIZE
                y2 = y1 + CELL_SIZE

                rid = self.canvas.create_rectangle(
                    x1, y1, x2, y2,
                    fill=COLORS["empty"], outline="", tags=f"cell_{r}_{c}"
                )
                self.rects[(r, c)] = rid
                self.cells[(r, c)] = "empty"

    def _get_cell(self, event):
        col = event.x // (CELL_SIZE + PADDING)
        row = event.y // (CELL_SIZE + PADDING)
        if 0 <= row < self.rows and 0 <= col < self.cols:
            return (row, col)
        return None

    def _on_click(self, event):
        self.is_drag = True
        cell = self._get_cell(event)
        if cell:
            self._apply_mode(cell)

    def _on_drag(self, event):
        if not self.is_drag:
            return
        cell = self._get_cell(event)
        if cell:
            mode = self.mode.get()
            if mode in ("obstacle", "clear"):
                self._apply_mode(cell)

    def _on_right_click(self, event):
        cell = self._get_cell(event)
        if cell:
            self._erase_cell(cell)

    def _on_right_drag(self, event):
        cell = self._get_cell(event)
        if cell:
            self._erase_cell(cell)

    def _apply_mode(self, cell):
        mode = self.mode.get()
        if mode == "start":
            if self.start and self.start != cell:
                self._set_cell(self.start, "empty")
            self.start = cell
            self._set_cell(cell, "start")

        elif mode == "end":
            if self.end and self.end != cell:
                self._set_cell(self.end, "empty")
            self.end = cell
            self._set_cell(cell, "end")

        elif mode == "obstacle":
            if self.cells[cell] not in ("start", "end"):
                self._set_cell(cell, "obstacle")

        elif mode == "clear":
            self._erase_cell(cell)

    def _erase_cell(self, cell):
        state = self.cells.get(cell, "empty")
        if state == "start":
            self.start = None
        elif state == "end":
            self.end = None
        self._set_cell(cell, "empty")

    def _set_cell(self, cell, state):
        self.cells[cell] = state
        self.canvas.itemconfig(self.rects[cell], fill=COLORS[state])

    def _set_mode(self, key):
        self.mode.set(key)
        for k in self.mode_buttons:
            self._update_btn_style(k)

    def _update_btn_style(self, key):
        btn = self.mode_buttons.get(key)
        if not btn:
            return
        active = (self.mode.get() == key)
        bg = COLORS["btn_act"] if active else COLORS["btn_bg"]
        fg = "#11111b" if active else COLORS["btn_text"]
        btn.configure(bg=bg, fg=fg)

    def _clear_grid(self):
        self._init_grid()

    def _clear_obstacles(self):
        for cell, state in list(self.cells.items()):
            if state == "obstacle":
                self._set_cell(cell, "empty")

    def _resize_grid(self):
        try:
            r = int(self._rows_var.get())
            c = int(self._cols_var.get())
            if not (3 <= r <= 40 and 3 <= c <= 60):
                raise ValueError
        except Exception:
            messagebox.showerror("Invalid Size",
                                 "Enter valid integers: rows 3-40, cols 3-60.")
            return

        self.rows = r
        self.cols = c
        
        max_cells = max(r, c)
        if max_cells <= 15:
            cell_size = 38
        elif max_cells <= 25:
            cell_size = 28
        elif max_cells <= 35:
            cell_size = 22
        else:
            cell_size = 18
        
        global CELL_SIZE
        CELL_SIZE = cell_size

        canvas_w = self.cols * (CELL_SIZE + PADDING) + PADDING
        canvas_h = self.rows * (CELL_SIZE + PADDING) + PADDING
        self.canvas.configure(width=canvas_w, height=canvas_h)

        self._init_grid()
        self.update()
    def _heuristic(self, cell1, cell2):
        if self._heuristic_var.get() == "manhattan":
            return abs(cell1[0] - cell2[0]) + abs(cell1[1] - cell2[1])
        else:
            import math
            return math.sqrt((cell1[0] - cell2[0])**2 + (cell1[1] - cell2[1])**2)

    def _get_neighbors(self, r, c):
        neighbors = []
        for dr, dc in ((-1,0),(0,1),(1,0),(0,-1)):
            nr, nc = r+dr, c+dc
            if 0 <= nr < self.rows and 0 <= nc < self.cols:
                neighbors.append((nr, nc))
        return neighbors

    def _astar_search(self):
        import heapq
        
        start = self.start
        end = self.end
        
        came_from = {start: None}
        g_score = {start: 0}
        f_score = {start: self._heuristic(start, end)}
        
        frontier = [(f_score[start], start)]
        explored = []
        queue_tracker = []
        
        while frontier:
            _, current = heapq.heappop(frontier)
            
            if current in explored:
                continue
            
            explored.append(current)
            queue_tracker.append(("explore", current))
            
            if current == end:
                break
            
            for neighbor in self._get_neighbors(*current):
                if self.cells.get(neighbor) == "obstacle":
                    continue
                
                tentative_g = g_score[current] + 1
                
                if neighbor not in g_score or tentative_g < g_score[neighbor]:
                    came_from[neighbor] = current
                    g_score[neighbor] = tentative_g
                    f_score[neighbor] = tentative_g + self._heuristic(neighbor, end)
                    heapq.heappush(frontier, (f_score[neighbor], neighbor))
                    queue_tracker.append(("queue", neighbor))
        
        path = []
        if end in came_from:
            current = end
            while current is not None:
                path.append(current)
                current = came_from.get(current)
            path.reverse()
        
        return queue_tracker, explored, path

    def _best_first_search_from(self, start_pos):
        import heapq
        
        end = self.end
        
        came_from = {start_pos: None}
        frontier = [(self._heuristic(start_pos, end), start_pos)]
        explored = []
        queue_tracker = []
        
        while frontier:
            _, current = heapq.heappop(frontier)
            
            if current in explored:
                continue
            
            explored.append(current)
            queue_tracker.append(("explore", current))
            
            if current == end:
                break
            
            for neighbor in self._get_neighbors(*current):
                if self.cells.get(neighbor) == "obstacle":
                    continue
                if neighbor not in came_from:
                    came_from[neighbor] = current
                    priority = self._heuristic(neighbor, end)
                    heapq.heappush(frontier, (priority, neighbor))
                    queue_tracker.append(("queue", neighbor))
        
        path = []
        if end in came_from:
            current = end
            while current is not None:
                path.append(current)
                current = came_from.get(current)
            path.reverse()
        
        return queue_tracker, explored, path

    def _astar_search_from(self, start_pos):
        import heapq
        
        end = self.end
        
        came_from = {start_pos: None}
        g_score = {start_pos: 0}
        f_score = {start_pos: self._heuristic(start_pos, end)}
        
        frontier = [(f_score[start_pos], start_pos)]
        explored = []
        queue_tracker = []
        
        while frontier:
            _, current = heapq.heappop(frontier)
            
            if current in explored:
                continue
            
            explored.append(current)
            queue_tracker.append(("explore", current))
            
            if current == end:
                break
            
            for neighbor in self._get_neighbors(*current):
                if self.cells.get(neighbor) == "obstacle":
                    continue
                
                tentative_g = g_score[current] + 1
                
                if neighbor not in g_score or tentative_g < g_score[neighbor]:
                    came_from[neighbor] = current
                    g_score[neighbor] = tentative_g
                    f_score[neighbor] = tentative_g + self._heuristic(neighbor, end)
                    heapq.heappush(frontier, (f_score[neighbor], neighbor))
                    queue_tracker.append(("queue", neighbor))
        
        path = []
        if end in came_from:
            current = end
            while current is not None:
                path.append(current)
                current = came_from.get(current)
            path.reverse()
        
        return queue_tracker, explored, path

    def _best_first_search(self):
        import heapq
        
        start = self.start
        end = self.end
        
        came_from = {start: None}
        frontier = [(self._heuristic(start, end), start)]
        explored = []
        queue_tracker = []
        
        while frontier:
            _, current = heapq.heappop(frontier)
            
            if current in explored:
                continue
            
            explored.append(current)
            queue_tracker.append(("explore", current))
            
            if current == end:
                break
            
            for neighbor in self._get_neighbors(*current):
                if self.cells.get(neighbor) == "obstacle":
                    continue
                if neighbor not in came_from:
                    came_from[neighbor] = current
                    priority = self._heuristic(neighbor, end)
                    heapq.heappush(frontier, (priority, neighbor))
                    queue_tracker.append(("queue", neighbor))
        
        path = []
        if end in came_from:
            current = end
            while current is not None:
                path.append(current)
                current = came_from.get(current)
            path.reverse()
        
        return queue_tracker, explored, path

    def _run_search(self):
        if self._sim_running or not self.start or not self.end:
            return
        
        self._clear_overlay()
        
        import time
        start_time = time.time()
        
        if self._algo_var.get() == "best_first":
            queue_tracker, explored, path = self._best_first_search()
        else:
            queue_tracker, explored, path = self._astar_search()
        
        end_time = time.time()
        execution_time_ms = (end_time - start_time) * 1000
        nodes_visited = len(explored)
        path_cost = len(path) - 1 if path else 0
        
        self._sim_running = True
        self._search_btn.configure(state="disabled")
        self._stop_btn.configure(state="normal")
        
        self._current_path = list(path)
        self._current_pos_index = 0
        self._replan_count = 0
        
        tracker_iter = iter(queue_tracker)
        
        def step():
            if not self._sim_running:
                return
            try:
                action, cell = next(tracker_iter)
            except StopIteration:
                if not path:
                    self._sim_running = False
                    self._search_btn.configure(state="normal")
                    self._stop_btn.configure(state="disabled")
                    messagebox.showerror(
                        "No Path Found",
                        "The algorithm could not find a path between start and end!\n\nThe end point may be completely surrounded by obstacles."
                    )
                    return
                delay = self._speed_var.get() * 3
                self._sim_after_id = self.after(delay, lambda: trace_path(iter(path)))
                return
            
            state = self.cells.get(cell)
            if state not in ("start", "end"):
                if action == "queue":
                    self._paint_cell(cell, COLORS["queue"])
                    self._overlays.add(cell)
                elif action == "explore":
                    self._paint_cell(cell, COLORS["visited"])
                    self._overlays.add(cell)
            
            self._sim_after_id = self.after(self._speed_var.get(), step)
        
        def trace_path(path_iter):
            if not self._sim_running:
                return
            
            if self._dynamic_var.get():
                import random
                spawn_prob = self._spawn_prob_var.get() / 100.0
                
                empty_cells = [
                    cell for cell in self.cells
                    if self.cells[cell] == "empty" and cell not in self._overlays
                    and cell != self.start and cell != self.end
                ]
                
                needs_replan = False
                for cell in empty_cells:
                    if random.random() < spawn_prob:
                        self._set_cell(cell, "obstacle")
                        self._paint_cell(cell, COLORS["dynamic_obstacle"])
                        
                        remaining_path = self._current_path[self._current_pos_index:]
                        if cell in remaining_path:
                            needs_replan = True
                
                if needs_replan:
                    current_cell = self._current_path[self._current_pos_index] if self._current_pos_index < len(self._current_path) else self.end
                    
                    if self._algo_var.get() == "best_first":
                        _, _, new_path = self._best_first_search_from(current_cell)
                    else:
                        _, _, new_path = self._astar_search_from(current_cell)
                    
                    if new_path:
                        self._replan_count += 1
                        self._current_path = new_path
                        self._current_pos_index = 0
                        
                        for cell in new_path[1:]:
                            if self.cells.get(cell) not in ("start", "end", "obstacle"):
                                self._paint_cell(cell, COLORS["path"])
                                self._overlays.add(cell)
                        
                        path_iter = iter(new_path[1:])
                    else:
                        self._sim_running = False
                        self._search_btn.configure(state="normal")
                        self._stop_btn.configure(state="disabled")
                        messagebox.showerror(
                            "Path Blocked!",
                            f"Dynamic obstacle blocked the path and no alternative route exists!\n\n"
                            f"Replans attempted: {self._replan_count}"
                        )
                        return
            
            try:
                cell = next(path_iter)
                self._current_pos_index += 1
            except StopIteration:
                self._paint_cell(self.start, COLORS["start"])
                self._paint_cell(self.end, COLORS["end"])
                self._sim_running = False
                self._search_btn.configure(state="normal")
                self._stop_btn.configure(state="disabled")
                
                algo_name = "Best-First Search" if self._algo_var.get() == "best_first" else "A* Search"
                dynamic_info = f"\n  • Replans: {self._replan_count}" if self._dynamic_var.get() else ""
                messagebox.showinfo(
                    "Path Found!",
                    f"✅ {algo_name} completed successfully!\n\n"
                    f"📊 Metrics:\n"
                    f"  • Nodes Visited: {nodes_visited}\n"
                    f"  • Path Cost: {path_cost} steps\n"
                    f"  • Execution Time: {execution_time_ms:.2f} ms{dynamic_info}"
                )
                return
            
            if self.cells.get(cell) not in ("start", "end"):
                self._paint_cell(cell, COLORS["agent"])
            
            self._sim_after_id = self.after(
                self._speed_var.get() * 2, lambda: trace_path(path_iter)
            )
        
        step()

    def _stop_sim(self):
        self._sim_running = False
        if self._sim_after_id:
            try:
                self.after_cancel(self._sim_after_id)
            except Exception:
                pass
            self._sim_after_id = None
        if hasattr(self, "_search_btn"):
            self._search_btn.configure(state="normal")
            self._stop_btn.configure(state="disabled")

    def _clear_overlay(self):
        self._stop_sim()
        for cell in list(self._overlays):
            self._paint_cell(cell, COLORS[self.cells.get(cell, "empty")])
        self._overlays.clear()
        if self.start and self.start in self.rects:
            self._paint_cell(self.start, COLORS["start"])
        if self.end and self.end in self.rects:
            self._paint_cell(self.end, COLORS["end"])

    def _paint_cell(self, cell, color):
        self.canvas.itemconfig(self.rects[cell], fill=color)

    def _generate_maze(self):
        import random
        
        density = self._density_var.get() / 100.0
        
        start_pos = self.start
        end_pos = self.end
        
        for r in range(self.rows):
            for c in range(self.cols):
                cell = (r, c)
                if cell == start_pos or cell == end_pos:
                    continue
                
                if random.random() < density:
                    self._set_cell(cell, "obstacle")
                else:
                    self._set_cell(cell, "empty")

if __name__ == "__main__":
    app = GridApp()
    app.mainloop()
```
